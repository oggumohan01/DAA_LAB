```python
import time

def coin_change(coins, amount):
    dp = [float('inf')] * (amount + 1)
    parent = [-1] * (amount + 1)

    dp[0] = 0

    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i and dp[i - coin] + 1 < dp[i]:
                dp[i] = dp[i - coin] + 1
                parent[i] = coin

    if dp[amount] == float('inf'):
        return None, []

    used = []
    x = amount

    while x > 0:
        used.append(parent[x])
        x -= parent[x]

    return dp[amount], used


# User Input
n = int(input("Enter number of coin denominations: "))

coins = list(map(int, input("Enter coin values: ").split()))

amount = int(input("Enter target amount: "))

start = time.perf_counter()

minimum, used_coins = coin_change(coins, amount)

end = time.perf_counter()

execution_time = end - start

# Output
print("\n------ RESULT ------")

if minimum is None:
    print("Change cannot be made.")
else:
    print("Minimum coins required:", minimum)
    print("Coins used:", used_coins)

print("Execution Time:", format(execution_time, ".10f"), "seconds")

print("\nTime Complexity: O(n × amount)")
print("Space Complexity: O(amount)")
```
