# Implementation of 0/1 Knapsack Problem using Dynamic Programming


```python


def knapsack(weights, profits, capacity):
    n = len(weights)

    # Create DP table
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    # Fill the DP table
    for i in range(1, n + 1):
        for w in range(1, capacity + 1):

            if weights[i - 1] <= w:
                include = profits[i - 1] + dp[i - 1][w - weights[i - 1]]
                exclude = dp[i - 1][w]

                dp[i][w] = max(include, exclude)
            else:
                dp[i][w] = dp[i - 1][w]

    return dp[n][capacity]


# Given weights and profits
weights = [1, 3, 4, 5]
profits = [1, 4, 5, 7]

# Knapsack capacity
capacity = 7

# Calculate maximum profit
maximum_profit = knapsack(weights, profits, capacity)

# Display the input data and result
print("Weights :", weights)
print("Profits :", profits)
print("Capacity:", capacity)
print("Maximum Profit =", maximum_profit)
```
