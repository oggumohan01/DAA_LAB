```python
import time

# Iterative method
def factorial_iterative(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact


# Recursive method
def factorial_recursive(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)


# User input
n = int(input("Enter a number: "))

if n < 0:
    print("Factorial is not defined for negative numbers.")

else:
    # Iterative execution
    start = time.perf_counter()
    result1 = factorial_iterative(n)
    end = time.perf_counter()

    iterative_time = end - start

    # Recursive execution
    start = time.perf_counter()
    result2 = factorial_recursive(n)
    end = time.perf_counter()

    recursive_time = end - start

    # Output
    print("\n----- RESULT -----")

    print("Number:", n)

    print("\nIterative Method")
    print("Factorial:", result1)
    print("Execution Time:",
          format(iterative_time, ".10f"), "seconds")

    print("\nRecursive Method")
    print("Factorial:", result2)
    print("Execution Time:",
          format(recursive_time, ".10f"), "seconds")

    print("\n----- COMPLEXITY -----")
    print("Iterative Time Complexity: O(n)")
    print("Iterative Space Complexity: O(1)")
    print("Recursive Time Complexity: O(n)")
    print("Recursive Space Complexity: O(n)")
```
